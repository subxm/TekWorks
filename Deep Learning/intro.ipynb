# Drawbacks of Machine Learning (ML)

## Where does Machine Learning fail?

Machine Learning often fails when **feature engineering becomes difficult or impossible**.

In traditional ML, humans must:
1. Identify important features from the raw data
2. Extract and prepare those features
3. Feed them into the model

If the correct features are not provided, the model performs poorly.

> ML fails not because algorithms are bad, but because feature engineering is difficult or impossible in many real-world problems.

---

# Core Idea of Deep Learning (DL)

Deep Learning solves this limitation by automatically learning features directly from raw data using neural networks.

> **Let the computer learn features by itself from raw data.**

---

# ML vs Deep Learning Workflow

## Machine Learning (ML)

```text
Raw Data → Feature Engineering → Model → Decision/Output
```

- Humans manually create features
- Performance depends heavily on feature quality

---

## Deep Learning (DL)

```text
Raw Data → Neural Network → Features → Decision/Output
```

- Features are learned automatically
- Minimal manual feature engineering required

---

# Drawbacks of Machine Learning / Deep Learning

## 1. Needs Large Amount of Data
Models require huge datasets for better accuracy.

## 2. High Computational Cost
Training models requires powerful hardware such as GPUs/TPUs.

## 3. Training Takes More Time
Complex models can take hours, days, or even weeks to train.

## 4. Overfitting Problem
Models may memorize training data instead of learning patterns.

## 5. Difficult to Understand Predictions (Black Box)
Many models, especially deep learning models, are hard to interpret.

## 6. Depends on Data Quality
Poor or biased data leads to poor predictions.

## 7. Can Produce Biased Results
If training data contains bias, the model also becomes biased.

## 8. Security and Privacy Risks
Sensitive data can be leaked or attacked.

## 9. Requires Continuous Maintenance
Models must be updated regularly with new data.

## 10. Does Not Think Like Humans
AI models recognize patterns but do not possess human understanding or reasoning.

# What is a Neuron?

A neuron is a **decision-making unit**.

It receives information, processes it, and decides whether to pass the signal forward or not.

Based on the processing result, the neuron decides:
- Pass the signal forward

# Components of an Artificial Neuron

## 1. Inputs

```text
x1, x2, x3, x4, .... xn
```

These are the values/data given to the neuron.

---

## 2. Weights

Weights determine the importance of each input.

```text
w1, w2, w3, .... wn
```

- Higher weight → More important input
- Lower weight → Less important input

---

## 3. Weighted Sum of Inputs

The neuron multiplies each input with its corresponding weight and adds them together.

```text
(x1*w1 + x2*w2 + x3*w3 + .... + xn*wn)
```

---

## 4. Bias

A bias value is added to shift the output.

```text
Weighted Sum + Bias
```

Bias helps the model learn better patterns.

---

## 5. Activation Function

The activation function decides whether the neuron should pass the signal/output forward or not.

```text
Output = Activation(Weighted Sum + Bias)
```

---

# Formula of an Artificial Neuron

```text
Output = Activation(
    x1*w1 + x2*w2 + x3*w3 + ... + xn*wn + bias
)
```

---

# Example

## Given:

```text
x1 = 9
x2 = 7
bias = -1
```

## Weighted Sum:

```text
sum = (x1*w1 + x2*w2 + bias)
```

Substituting values:

```text
sum = (9*w1 + 7*w2 - 1)
```

The activation function then processes this sum to produce the final output.

---

# Flow of an Artificial Neuron

```text
Inputs → Weights → Sum → Bias → Activation Function → Output
```

# Perceptron

A perceptron is a **single neuron model** used in Deep Learning.

```text
Inputs → Weights → Sum → Activation Function → Output
```

---

# Limitation of Perceptron

The main limitation of a perceptron is:

> It is only limited to binary classification problems.

This means it can classify only two classes such as:
- Yes / No
- Spam / Not Spam
- Pass / Fail

---

# Activation Functions

Activation functions help the neuron decide whether to activate/pass the signal forward.

Without an activation function, the neuron output remains only linear.

```text
x1*w1 + x2*w2 + b  → Linear Output
```

Activation functions introduce non-linearity, allowing Deep Learning models to solve complex problems.

---

# Types of Activation Functions

## 1. Sigmoid Activation Function

The sigmoid function gives output values between `0` and `1`.

### Decision:
- Output close to `1` → Yes / True
- Output close to `0` → No / False

### Examples:
- Spam or Not Spam
- Pass or Fail

### Sigmoid Formula

```text
σ(x) = 1 / (1 + e^-x)
```

---

## 2. ReLU Activation Function

ReLU stands for:

```text
Rectified Linear Unit
```

### Rules:

```text
If value < 0  → Output = 0
If value > 0  → Output = value
```

### ReLU Formula

```text
f(x) = max(0, x)
```

ReLU is widely used in hidden layers of Deep Learning models.

---

## 3. Softmax Activation Function

Softmax is used when there are **multiple classes**.

### Use Case:
Multi-class classification problems.

### Examples:
- Cat / Dog / Horse
- Apple / Banana / Mango
- Handwritten digit recognition (0–9)

Softmax converts outputs into probabilities.

---

# Summary Table

| Activation Function | Usage | Output |
|---|---|---|
