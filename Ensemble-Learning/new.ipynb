# Bagging and Boosting

# Ensemble Learning

Ensemble learning combines multiple machine learning models to improve performance, accuracy, and robustness.

Main types:
1. Bagging (Bootstrap Aggregating)
2. Boosting

---

# 1. Bagging (Bootstrap Aggregating)

## Definition

Bagging is an ensemble technique where multiple models are trained independently on different random subsets of the training data, and their outputs are combined.

Goal:
- Reduce variance
- Avoid overfitting
- Improve stability

---

# Working of Bagging

## Steps of Bagging

1. Take the original dataset.
2. Create multiple random subsets using Bootstrap Sampling.
   - Sampling is done with replacement.
3. Train a separate model on each subset.
4. Each model predicts independently.
5. Combine predictions:
   - Classification → Majority Voting
   - Regression → Averaging

---

# Bootstrap Sampling

Bootstrap sampling means:
- Randomly selecting samples from dataset
- Selection is done with replacement

Example:

Original Data:
```text
[1, 2, 3, 4, 5]
```

Bootstrap Sample:
```text
[2, 4, 4, 1, 5]
```

Some values may repeat, some may be missing.

---

# Advantages of Bagging

- Reduces overfitting
- Reduces variance
- Improves accuracy
- Works well with high variance models
- Parallel training possible

---

# Disadvantages of Bagging

- Increased computation
- Less interpretable
- Does not reduce bias much

---

# Random Forest

## Definition

Random Forest is a Bagging-based ensemble algorithm that uses multiple Decision Trees.

It combines:
- Bootstrap Sampling
- Random Feature Selection

---

# Why Random Forest?

Single Decision Trees:
- Overfit easily
- High variance

Random Forest:
- Combines many trees
- Produces more stable predictions

---

# Working of Random Forest

## Steps of Random Forest

### Step 1: Bootstrap Sampling

Create multiple subsets from original data.

---

### Step 2: Train Multiple Decision Trees

For each tree:
- Random subset of rows is selected
- Random subset of features is selected

---

### Step 3: Independent Training

Each tree trains independently.

---

### Step 4: Prediction

For Classification:
- Majority Voting

For Regression:
- Average Prediction

---

# Random Feature Selection

At every split:
- Random subset of features is considered
- Not all features are checked

This:
- Reduces correlation between trees
- Improves diversity

---

# Random Forest Architecture

```text
Dataset
   ↓
Bootstrap Samples
   ↓
Multiple Decision Trees
   ↓
Predictions from each tree
   ↓
Voting / Averaging
   ↓
Final Output
```

---

# Important Parameters in Random Forest

| Parameter | Meaning |
|---|---|
| n_estimators | Number of trees |
| max_depth | Maximum depth of tree |
| min_samples_split | Minimum samples to split |
| min_samples_leaf | Minimum samples in leaf |
| max_features | Number of features considered |
| bootstrap | Whether bootstrap sampling is used |

---

# Out Of Bag (OOB) Error

Some samples are not selected during bootstrap sampling.

These are called:
- Out Of Bag Samples

Used for:
- Internal validation
- Estimating model accuracy

No separate validation set needed.

---

# Advantages of Random Forest

- High accuracy
- Reduces overfitting
- Handles missing values
- Handles large datasets
- Feature importance available
- Works for classification and regression

---

# Disadvantages of Random Forest

- Slower than single tree
- Large memory usage
- Less interpretable
- Many trees increase computation

---

# Applications of Random Forest

- Fraud Detection

- Recommendation Systems

# 2. Boosting

## Definition

Boosting is an ensemble technique where models are trained sequentially, and each new model focuses on correcting previous model errors.

Goal:
- Reduce bias
- Improve prediction accuracy

---

# Working of Boosting

## Steps of Boosting

1. Train first weak learner.
2. Identify wrongly predicted samples.
3. Increase importance (weight) of wrong predictions.
4. Train next model focusing more on errors.
5. Repeat sequentially.
6. Combine all weak learners into strong learner.

---

# Weak Learner

A weak learner:
- Performs slightly better than random guessing

Example:
- Decision Stump (tree with depth = 1)

---

# Bagging vs Boosting

| Feature | Bagging | Boosting |
|---|---|---|
| Training | Parallel | Sequential |
| Goal | Reduce Variance | Reduce Bias |
| Model Dependency | Independent | Dependent |
| Overfitting | Less | Can overfit |
| Speed | Faster | Slower |
| Example | Random Forest | AdaBoost |

---

# AdaBoost (Adaptive Boosting)

## Definition

AdaBoost is a boosting algorithm that combines multiple weak learners sequentially.

Each new learner focuses more on previously misclassified samples.

Usually uses:
- Decision Stumps

---

# Core Idea of AdaBoost

- Wrong predictions get higher weight
- Correct predictions get lower weight
- Next model focuses on difficult samples

---

# Working of AdaBoost

## Step-by-Step Process

---

## Step 1: Initialize Weights

All training samples receive equal weights.

If dataset has N samples:

```math
w_i = \frac{1}{N}
```

---

## Step 2: Train Weak Learner

Train first Decision Stump on weighted data.

---

## Step 3: Calculate Error

Compute weighted error:

```math
Error = \sum w_i \text{(misclassified samples)}
```

---

## Step 4: Calculate Model Weight

Calculate importance of weak learner:

```math
\alpha = \frac{1}{2}\ln\left(\frac{1-Error}{Error}\right)
```

Where:
- Smaller error → Larger alpha
- Larger alpha → More important model

---

## Step 5: Update Sample Weights

Increase weights for wrongly classified samples.

Update rule:

```math
w_i \leftarrow w_i e^{\alpha}
```

Normalize weights afterward.

---

## Step 6: Train Next Weak Learner

New learner focuses more on difficult samples.

---

## Step 7: Repeat

Repeat until:
- Fixed number of learners
- Desired accuracy reached

---

## Step 8: Final Prediction

Combine all weak learners using weighted voting.

---

# AdaBoost Architecture

```text
Dataset
   ↓
Weak Learner 1
   ↓
Increase weights of wrong predictions
   ↓
Weak Learner 2
   ↓
Increase weights again
   ↓
Weak Learner 3
   ↓
Weighted Combination
   ↓
Final Strong Learner
```

---
