# K-Nearest Neighbors (KNN)

**Simple, intuitive, instance-based supervised learning algorithm** used for both **classification** and **regression**.

---

## How KNN Works

1. Store all training examples (lazy learner - no explicit training phase).
2. For a new data point:
   - Calculate distance to every point in training set.
   - Find the **K** nearest neighbors.
   - **Classification**: Majority vote of neighbors.
   - **Regression**: Average (or weighted average) of neighbors' values.

---

## Distance Metrics

| Metric | Formula | Best For |
|--------|--------|----------|
| Euclidean | `√Σ(xi - yi)²` | Continuous data |
| Manhattan | `Σ|xi - yi|` | High dimensions, grid-like |
| Minkowski | Generalized | Flexible |
| Hamming | Number of differing positions | Categorical |
| Cosine | 1 - (A·B)/(‖A‖‖B‖) | Text, high-dimensional sparse data |

---

## Choosing K

- **Small K** → Low bias, high variance (sensitive to noise).
- **Large K** → High bias, low variance (smoother decision boundary).
- Common choices: 3, 5, 7, or √N (N = training samples).
- Use **cross-validation** to find optimal K.
